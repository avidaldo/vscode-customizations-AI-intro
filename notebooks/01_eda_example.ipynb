## 1. Load and Inspect

We load the CSV into a pandas DataFrame and run a quick structural check: shape, column types, missing values.

```
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

DATA_PATH = Path("../data/sample.csv")
df = pd.read_csv(DATA_PATH)

print(f"Shape: {df.shape}")
print(f"Columns: {list(df.columns)}")
df.head()
```

## 2. Missing Values and Basic Statistics

We check for null values and inspect the basic descriptive statistics for each numeric feature.

```
print("Missing values per column:")
print(df.isnull().sum())
print()
print("Class distribution:")
print(df["species"].value_counts())
print()
df.describe()
```

## 3. Feature Distributions by Class

Box plots let us visually assess how well each feature separates the three Iris species.
Features with higher inter-class variance are likely to be more informative for a classifier.

```
features = ["sepal_length", "sepal_width", "petal_length", "petal_width"]

fig, axes = plt.subplots(1, 4, figsize=(14, 4))
for ax, feat in zip(axes, features):
    df.boxplot(column=feat, by="species", ax=ax)
    ax.set_title(feat)
    ax.set_xlabel("")
plt.suptitle("Feature distributions by species")
plt.tight_layout()
plt.show()
```

**Interpretation:** `petal_length` and `petal_width` show the clearest separation between *setosa* and the other two species. *versicolor* and *virginica* overlap more in sepal measurements, suggesting that petal features will be more discriminative for a classifier.

## 4. Pairplot

A pairplot shows all pairwise scatter plots coloured by class, revealing which feature combinations best separate the three species.

```
sns.pairplot(df, hue="species", plot_kws={"alpha": 0.7})
plt.suptitle("Pairwise feature scatter by species", y=1.02)
plt.show()
```

**Interpretation:** *setosa* is linearly separable from the other two species across almost all feature combinations. *versicolor* and *virginica* show more overlap, particularly in sepal dimensions, which motivates using all four features in a classifier rather than a subset.

## 5. Summary and Next Steps

**Key findings:**
- The dataset has no missing values and is well-structured.
- 3 classes with 20, 10, and 10 samples respectively — the sample is imbalanced; a full Iris dataset has 50 per class.
- `petal_length` and `petal_width` are the most discriminative features.
- *setosa* is linearly separable; *versicolor* / *virginica* require more expressive models.
